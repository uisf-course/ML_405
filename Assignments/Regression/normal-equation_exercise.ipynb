## 1. Theory: Normal Equation

Given a design matrix $X$ and target vector $y$, we want to find parameters $\theta$ such that the predictions

$$
\hat{y} = X\theta
$$

are as close as possible to $y$.

Starting from the normal-equation derivation:

$$
X^T X\theta = X^T y
$$

we obtain

$$
\boxed{\theta = (X^T X)^{-1}X^T y}
$$

In practice, $X^T X$ may not be invertible. Following the reference notebook, we will use the **pseudo-inverse** instead:

$$
\boxed{\theta = (X)^{+}y}
$$

where $(\cdot)^+$ denotes the pseudo-inverse.

For one input feature $x$, a linear model is

$$
h_\theta(x) = \theta_0 + \theta_1x.
$$

Therefore each sample must be represented as $[1, x]$. The first column of ones is the **bias/intercept column**.

# Part 2 — Polynomial Regression

A straight line cannot represent every relationship. Polynomial regression keeps the model **linear in the parameters** while adding nonlinear transformations of the input.

For degree $d$:

$$
h_\theta(x)=\theta_0+\theta_1x+\theta_2x^2+\cdots+\theta_dx^d.
$$

The corresponding design matrix becomes

$$
X_{poly} =
\begin{bmatrix}
1 & x_1 & x_1^2 & \cdots & x_1^d \\
1 & x_2 & x_2^2 & \cdots & x_2^d \\
\vdots & \vdots & \vdots & & \vdots \\
1 & x_n & x_n^2 & \cdots & x_n^d
\end{bmatrix}.
$$

### Important idea

**The normal-equation function does not change.** Only the design matrix changes.

That means you can reuse the same `normal_equation(X, y)` function after generating polynomial features.
