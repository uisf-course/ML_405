## Setup

This notebook covers the Pandas tools that are most useful when preparing and exploring data for machine learning.

Pandas is mainly used for working with **structured data** such as tables and CSV files.

````python
import numpy as np
import pandas as pd
````

## Series and DataFrames

### Series

A Series is a one-dimensional labeled data structure.

A single column of a dataset is usually represented as a Series.

````python
scores = pd.Series([65, 82, 58, 91, 85])

scores
````

### DataFrame

A `DataFrame` is a two-dimensional table with rows and columns.

For an ML dataset:

- rows usually represent **samples**
- columns usually represent **features**
- one column may represent the **target**

````python
data = {
    "age": [20, 21, 19, 22, 20, 23, 21, 19, 24, 22,
            20, 25, 21, 23, 19, 22, 24, 20, 21, 23],

    "hours_studied": [3, 5, 2, 7, 6, 8, 4, 2, 9, 6,
                      3, 10, 5, 7, 1, 6, 8, 4, 5, 7],

    "attendance": [70, 85, 65, 95, 90, 98, 78, 60, 92, 88,
                   72, 96, 84, 91, 55, 87, 94, 76, 82, 89],

    "score": [65, 82, 58, 91, 85, 96, 73, 52, 94, 87,
              68, 98, 80, 90, 45, 84, 92, 70, 79, 88]
}

df = pd.DataFrame(data)

df
````

## Loading a Dataset

Pandas provides functions for reading data from different file formats. In ML, CSV files are especially common, but you may also encounter Excel and plain text files.

General syntax:

```python
df = pd.read_csv("data.csv")
df = pd.read_excel("data.xlsx")
```
For a text file, the appropriate function depends on how the data is structured.

if values are separated by spaces:
```python
df = pd.read_csv("data.txt", sep=" ")
```

If values are separated by tabs:
```python
df = pd.read_csv("data.txt", sep="\t")
```

If values are separated by commas:
```python
df = pd.read_csv("data.txt", sep=",")
```

The result is a Pandas DataFrame.

Before training a model, we usually inspect the dataset to understand its size, columns, data types, and basic statistics.

## Indexing and Selection

### Selecting One Column

````python
df["hours_studied"]
````

Selecting one column with one pair of brackets returns a Series.

### Selecting Multiple Columns

````python
df[["hours_studied", "attendance", "score"]]
````

Selecting multiple columns requires a list of column names.

This returns a DataFrame.

### iloc

iloc selects data using integer positions.

General syntax:

```python
df.iloc[rows, columns]
```

````python
# First row
df.iloc[0]
````

````python
# First three rows
df.iloc[:3]
````

````python
# First three rows and first two columns
df.iloc[:3, :2]
````

### loc

loc selects data using labels or Boolean conditions.

````python
# Rows where age is 20:
df.loc[df["age"] == 20]
````

````python
df.loc[:, ["hours_studied", "score"]]
````

### Filtering Rows

````python
df[df["score"] >= 80]
````

Multiple conditions can be combined with:

- `&` for AND
- `|` for OR

Each condition should be inside parentheses.

````python
df[
    (df["hours_studied"] >= 5) &
    (df["score"] >= 80)
]
````

## Working with Missing Values

Real datasets often contain missing values.

Pandas represents many missing numerical values as `NaN`.

````python
df_missing = pd.DataFrame({
    "hours_studied": [3, 5, np.nan, 7, 6],
    "attendance": [70, 85, 65, np.nan, 90],
    "score": [65, 82, 58, 91, 85]
})

df_missing
````

### Detecting Missing Values

````python
df_missing.isna()
````

````python
df_missing.isna().sum()
````

### Removing Missing Values

````python
df_missing.dropna()
````

### Filling Missing Values

````python
df_missing.fillna(0)
````

## Basic Column Operations

Pandas can perform vectorized calculations on complete columns.

````python
df["score_fraction"] = df["score"] / 100

df
````

We can also create Boolean columns from conditions.

````python
df["passed"] = df["score"] >= 60

df
````

If a temporary column is no longer needed, drop() can remove it.

````python
df = df.drop(columns=["score_fraction"])

df
````

## Correlation

Correlation measures the strength and direction of a linear relationship between two numerical variables.

Its value is between -1 and +1:

- +1  -> strong positive linear relationship
-  0  -> little/no linear relationship
- -1  -> strong negative linear relationship

````python
df["hours_studied"].corr(df["score"])
````

````python
# correlation matrix for all numerical columns.
df.corr(numeric_only=True)
````

## Separating Features and Target

One of the most common steps in an ML workflow is separating the input features from the target.

By convention:

- X → features
- y → target

````python
X = df[["age", "hours_studied", "attendance"]]
y = df["score"]
print("X:\n", X)
print("\nX shape:", X.shape)
print("y shape:", y.shape)
````

## Converting Pandas Data to NumPy

Many ML libraries can work directly with Pandas, but converting to NumPy is sometimes useful. for this matter we use to_numpy().

````python
X_array = X.to_numpy()
y_array = y.to_numpy()

print(type(X_array))
print(type(y_array))

print("\nX_array:")
print(X_array)
````
